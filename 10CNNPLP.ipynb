```python
# Load various imports
from datetime import datetime
from os import listdir
from os.path import isfile, join
import librosa
import librosa.display
import numpy as np
import pandas as pd
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Conv2D, MaxPooling2D, GlobalAveragePooling2D, Input
from keras.utils import to_categorical
from tensorflow.keras.callbacks import ModelCheckpoint
from sklearn.metrics import confusion_matrix, classification_report, roc_curve, auc
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt
import seaborn as sns

# Path to the audio files
mypath = "E:\\FFF\\respiratory-sound-database\\Respiratory_Sound_Database\\Respiratory_Sound_Database\\audio_and_txt_files\\"
filenames = [f for f in listdir(mypath) if (isfile(join(mypath, f)) and f.endswith('.wav'))]

# Extract patient IDs from filenames
p_id_in_file = [int(name[:3]) for name in filenames]
p_id_in_file = np.array(p_id_in_file)

# Maximum padding length for consistent feature size
max_pad_len = 862

def extract_features(file_name):
    """
    This function takes in the path for an audio file as a string, loads it, and returns the PLP features.
    """
    try:
        audio, sample_rate = librosa.load(file_name, res_type='kaiser_fast', duration=20)
        # Compute PLP features using librosa
        plp = librosa.feature.melspectrogram(y=audio, sr=sample_rate, n_mels=40)
        plp = librosa.power_to_db(plp, ref=np.max)  # Convert to dB scale
        pad_width = max_pad_len - plp.shape[1]
        plp = np.pad(plp, pad_width=((0, 0), (0, pad_width)), mode='constant')
    except Exception as e:
        print("Error encountered while parsing file: ", file_name)
        return None
    return plp

# Full paths of files
filepaths = [join(mypath, f) for f in filenames]

# Load patient diagnosis file
p_diag = pd.read_csv(".\\respiratory-sound-database\\Respiratory_Sound_Database\\Respiratory_Sound_Database\\patient_diagnosis.csv", header=None)

# Labels for audio files
labels = np.array([p_diag[p_diag[0] == x][1].values[0] for x in p_id_in_file])

# Extract features
features = []
for file_name in filepaths:
    data = extract_features(file_name)
    if data is not None:
        features.append(data)

print('Finished feature extraction from ', len(features), ' files')
features = np.array(features)  # Shape: (num_files, num_mels, max_pad_len)

# Plot PLP features
plt.figure(figsize=(10, 4))
librosa.display.specshow(features[7], x_axis='time', y_axis='mel')
plt.colorbar(format='%+2.0f dB')
plt.title('Perceptual Linear Prediction (PLP)')
plt.tight_layout()
plt.show()

# Delete rare diseases
features1 = np.delete(features, np.where((labels == 'Asthma') | (labels == 'LRTI'))[0], axis=0)
labels1 = np.delete(labels, np.where((labels == 'Asthma') | (labels == 'LRTI'))[0], axis=0)

# Print class counts
unique_elements, counts_elements = np.unique(labels1, return_counts=True)
print(np.asarray((unique_elements, counts_elements)))

# Plot class counts
y_pos = np.arange(len(unique_elements))
plt.figure(figsize=(12, 8))
plt.bar(unique_elements, counts_elements, align='center', alpha=0.5)
plt.xticks(y_pos, unique_elements)
plt.ylabel('Count')
plt.xlabel('Disease')
plt.title('Disease Count in Sound Files (No Asthma or LRTI)')
plt.show()

# One-hot encode labels
le = LabelEncoder()
i_labels = le.fit_transform(labels1)
oh_labels = to_categorical(i_labels)

# Add channel dimension for CNN
features1 = np.reshape(features1, (features1.shape[0], features1.shape[1], features1.shape[2], 1))  # Shape: (num_files, num_mels, max_pad_len, 1)

# Train-test split
x_train, x_test, y_train, y_test = train_test_split(features1, oh_labels, stratify=oh_labels, test_size=0.2, random_state=42)

# Model parameters
num_rows = features1.shape[1]  # Number of Mel bands
num_columns = max_pad_len  # Time steps
num_channels = 1  # Single channel
num_labels = oh_labels.shape[1]
filter_size = 2  # Kernel size

# Construct model
model = Sequential()
model.add(Input(shape=(num_rows, num_columns, num_channels)))  # Explicit Input layer
model.add(Conv2D(filters=16, kernel_size=filter_size, activation='relu'))
model.add(MaxPooling2D(pool_size=2))
model.add(Dropout(0.2))

model.add(Conv2D(filters=32, kernel_size=filter_size, activation='relu'))
model.add(MaxPooling2D(pool_size=2))
model.add(Dropout(0.2))

model.add(Conv2D(filters=64, kernel_size=filter_size, activation='relu'))
model.add(MaxPooling2D(pool_size=2))
model.add(Dropout(0.2))

model.add(Conv2D(filters=128, kernel_size=filter_size, activation='relu'))
model.add(MaxPooling2D(pool_size=2))
model.add(Dropout(0.2))

model.add(GlobalAveragePooling2D())
model.add(Dense(num_labels, activation='softmax'))

# Compile the model
model.compile(loss='categorical_crossentropy', metrics=['accuracy'], optimizer='adam')

# Display model architecture summary
model.summary()

# Calculate pre-training accuracy
score = model.evaluate(x_test, y_test, verbose=1)
accuracy = 100 * score[1]
print("Pre-training accuracy: %.4f%%" % accuracy)

# Train model
num_epochs = 250
num_batch_size = 128

callbacks = [
    ModelCheckpoint(
        filepath='mymodel2_{epoch:02d}.keras',
        save_best_only=True,
        monitor='val_accuracy',
        verbose=1)
]

start = datetime.now()
model.fit(x_train, y_train, batch_size=num_batch_size, epochs=num_epochs, validation_data=(x_test, y_test), callbacks=callbacks, verbose=1)
duration = datetime.now() - start
print("Training completed in time: ", duration)

# Evaluate the model
score = model.evaluate(x_train, y_train, verbose=0)
print("Training Accuracy: ", score[1])

score = model.evaluate(x_test, y_test, verbose=0)
print("Testing Accuracy: ", score[1])

# Predictions
preds = model.predict(x_test)
classpreds = np.argmax(preds, axis=1)
y_testclass = np.argmax(y_test, axis=1)

# ROC Curve and AUC
n_classes = 6
fpr = dict()
tpr = dict()
roc_auc = dict()
for i in range(n_classes):
    fpr[i], tpr[i], _ = roc_curve(y_test[:, i], preds[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

# Class names
c_names = ['Bronchiectasis', 'Bronchiolitis', 'COPD', 'Healthy', 'Pneumonia', 'URTI']

# Plot ROC curves
fig, ax = plt.subplots(figsize=(16, 10))
ax.plot([0, 1], [0, 1], 'k--')
ax.set_xlim([0.0, 1.0])
ax.set_ylim([0.0, 1.05])
ax.set_xlabel('False Positive Rate')
ax.set_ylabel('True Positive Rate')
ax.set_title('ROC Curve for Each Class')
for i in range(n_classes):
    ax.plot(fpr[i], tpr[i], linewidth=3, label='ROC curve (area = %0.2f) for %s' % (roc_auc[i], c_names[i]))
ax.legend(loc="best", fontsize='x-large')
ax.grid(alpha=.4)
sns.despine()
plt.show()

# Classification Report
print(classification_report(y_testclass, classpreds, target_names=c_names))

# Confusion Matrix
print(confusion_matrix(y_testclass, classpreds))
```

```python
# ======================
# 1. IMPORTS & CONFIGURATION
# ======================
import os
import numpy as np
import pandas as pd
import librosa
from scipy.fftpack import dct
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (Conv2D, MaxPooling2D, 
                                   GlobalAveragePooling2D, 
                                   Dense, Dropout, Input)
from tensorflow.keras.utils import to_categorical
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
import matplotlib.pyplot as plt

# ======================
# 2. CONFIGURATION (YOUR PATHS)
# ======================
AUDIO_PATH = r"F:\FFF\respiratory-sound-database\Respiratory_Sound_Database\Respiratory_Sound_Database\audio_and_txt_files"
DIAGNOSIS_PATH = r"F:\FFF\respiratory-sound-database\Respiratory_Sound_Database\Respiratory_Sound_Database\patient_diagnosis.csv"

# Audio parameters
SAMPLE_RATE = 22050
DURATION = 20  # seconds

# PLP parameters
WINLEN = 0.025    # 25ms window
WINSTEP = 0.01    # 10ms step
NUMCEP = 13       # PLP coefficients
NFILT = 26        # Mel filters
MAX_PAD_LEN = 862 # Time steps

# ======================
# 3. FIXED PLP EXTRACTION
# ======================
def extract_plp_features(file_path):
    """Robust PLP feature extraction with automatic FFT sizing"""
    try:
        # Load audio with padding if shorter than DURATION
        audio, sr = librosa.load(file_path,
                               sr=SAMPLE_RATE,
                               duration=DURATION,
                               res_type='kaiser_fast')
        
        # Calculate frame parameters
        frame_length = int(WINLEN * SAMPLE_RATE)
        hop_length = int(WINSTEP * SAMPLE_RATE)
        
        # Compute STFT with automatic sizing
        stft = librosa.stft(audio,
                          win_length=frame_length,
                          hop_length=hop_length,
                          center=True,
                          pad_mode='reflect')
        
        # Get actual FFT size used
        n_fft = 2 * (stft.shape[0] - 1)
        
        # Power spectrum
        power_spec = np.abs(stft)**2
        
        # Mel filterbank with correct size
        mel_basis = librosa.filters.mel(sr=SAMPLE_RATE,
                                      n_fft=n_fft,
                                      n_mels=NFILT)
        
        # Apply filterbank
        mel_energy = np.dot(mel_basis, power_spec)
        
        # Log compression and DCT
        log_mel = np.log(mel_energy + 1e-6)
        plp_coeffs = dct(log_mel, type=2, axis=0, norm='ortho')[:NUMCEP]
        
        # Pad/truncate time dimension
        if plp_coeffs.shape[1] > MAX_PAD_LEN:
            return plp_coeffs[:, :MAX_PAD_LEN]
        return np.pad(plp_coeffs,
                    ((0, 0), (0, MAX_PAD_LEN - plp_coeffs.shape[1])),
                    mode='constant')
    
    except Exception as e:
        print(f"Error processing {os.path.basename(file_path)}: {str(e)}")
        return None

# ======================
# 4. DATA PROCESSING
# ======================
print("=== Loading Data ===")
filenames = [f for f in os.listdir(AUDIO_PATH) if f.endswith('.wav')]
print(f"Found {len(filenames)} files")

# Load diagnoses
diagnoses = pd.read_csv(DIAGNOSIS_PATH, header=None)
p_ids = np.array([int(f[:3]) for f in filenames])
labels = np.array([diagnoses[diagnoses[0] == pid][1].values[0] for pid in p_ids])

# Feature extraction
print("\n=== Extracting Features ===")
features = []
valid_files = []

for i, fname in enumerate(filenames):
    filepath = os.path.join(AUDIO_PATH, fname)
    feat = extract_plp_features(filepath)
    if feat is not None:
        features.append(feat)
        valid_files.append(fname)
    if (i+1) % 50 == 0:
        print(f"Processed {i+1}/{len(filenames)} files")

features = np.array(features)
labels = labels[[i for i, f in enumerate(filenames) if f in valid_files]]

# Filter classes
keep_mask = ~np.isin(labels, ['Asthma', 'LRTI'])
features = features[keep_mask]
labels = labels[keep_mask]

print("\n=== Dataset Summary ===")
print(f"Final samples: {features.shape[0]}")
print(f"Feature shape: {features.shape[1:]}")
print("Class distribution:")
print(pd.Series(labels).value_counts())

# ======================
# 5. MODEL TRAINING
# ======================
# Encode labels
le = LabelEncoder()
y = to_categorical(le.fit_transform(labels))

# Reshape for CNN
X = np.expand_dims(features, axis=-1)

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)

# Build model
model = Sequential([
    Input(shape=X.shape[1:]),
    Conv2D(32, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Dropout(0.3),
    Conv2D(64, (3, 3), activation='relu'),
    GlobalAveragePooling2D(),
    Dense(y.shape[1], activation='softmax')
])
model.compile(optimizer='adam',
             loss='categorical_crossentropy',
             metrics=['accuracy'])

# Training
print("\n=== Training ===")
history = model.fit(X_train, y_train,
                   validation_data=(X_test, y_test),
                   epochs=50,
                   batch_size=128,
                   verbose=1)

# ======================
# 6. EVALUATION
# ======================
# Classification report
y_pred = np.argmax(model.predict(X_test), axis=1)
y_true = np.argmax(y_test, axis=1)
print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=le.classes_))

# Visualization
plt.figure(figsize=(12, 4))
librosa.display.specshow(features[0],
                        x_axis='time',
                        sr=SAMPLE_RATE,
                        hop_length=int(WINSTEP * SAMPLE_RATE))
plt.colorbar()
plt.title('PLP Features (First Sample)')
plt.show()
```
