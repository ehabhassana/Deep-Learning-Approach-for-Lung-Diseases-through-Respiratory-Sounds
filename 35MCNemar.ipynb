```python
import pandas as pd
import numpy as np
from statsmodels.stats.contingency_tables import mcnemar

# Load predictions from both experiments
df_smote = pd.read_csv('E:\\Results\\exp35_smote_mcnemar\\predictions.csv')
df_cw = pd.read_csv('E:\\Results\\exp35_classweight_mcnemar\\predictions.csv')

# Ensure same test samples (should be same due to same random_state in split)
assert len(df_smote) == len(df_cw), "Different number of test samples!"
assert (df_smote['y_true'] == df_cw['y_true']).all(), "y_true mismatch!"

y_true = df_smote['y_true']
y_pred_smote = df_smote['y_pred']
y_pred_cw = df_cw['y_pred']

# Build contingency table
a = np.sum((y_pred_smote == y_true) & (y_pred_cw == y_true))
b = np.sum((y_pred_smote == y_true) & (y_pred_cw != y_true))
c = np.sum((y_pred_smote != y_true) & (y_pred_cw == y_true))
d = np.sum((y_pred_smote != y_true) & (y_pred_cw != y_true))

table = [[a, b], [c, d]]
print("Contingency table:")
print(table)

# McNemar test (exact)
result = mcnemar(table, exact=True)
print('p-value:', result.pvalue)

if result.pvalue < 0.05:
    print("There is a statistically significant difference between the two models.")
else:
    print("No significant difference.")
```
