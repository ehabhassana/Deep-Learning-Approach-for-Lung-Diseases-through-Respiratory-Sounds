```python
import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.manifold import TSNE
import umap  # 
import numpy as np
from os import listdir
from os.path import isfile, join
import librosa
import pandas as pd
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Conv1D, MaxPooling1D, GlobalAveragePooling1D, Input
from tensorflow.keras.utils import to_categorical
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt
import seaborn as sns

# File paths and loading data
mypath = "F:\\FFF\\respiratory-sound-database\\Respiratory_Sound_Database\\Respiratory_Sound_Database\\audio_and_txt_files\\"
filenames = [f for f in listdir(mypath) if (isfile(join(mypath, f)) and f.endswith('.wav'))]
p_id_in_file = [int(name[:3]) for name in filenames]
p_id_in_file = np.array(p_id_in_file)
max_pad_len = 862

# Feature extraction function
def extract_features(file_name):
    try:
        audio, sample_rate = librosa.load(file_name, res_type='kaiser_fast', duration=20)
        energy = librosa.feature.rms(y=audio)
        energy_mean = np.mean(energy, axis=1)
        spectral_bandwidth = librosa.feature.spectral_bandwidth(y=audio, sr=sample_rate)
        spectral_bandwidth_mean = np.mean(spectral_bandwidth, axis=1)
        mfccs = librosa.feature.mfcc(y=audio, sr=sample_rate, n_mfcc=13)
        mfccs_mean = np.mean(mfccs, axis=1)
        features = np.hstack([energy_mean, spectral_bandwidth_mean, mfccs_mean])
        pad_width = max_pad_len - features.shape[0]
        if pad_width > 0:
            features = np.pad(features, pad_width=(0, pad_width), mode='constant')
        elif pad_width < 0:
            features = features[:max_pad_len]
        return features
    except Exception as e:
        print(f"Error parsing {file_name}: {e}")
        return None

# Extract features and labels
filepaths = [join(mypath, f) for f in filenames]
features = np.array([extract_features(f) for f in filepaths if extract_features(f) is not None])
p_diag = pd.read_csv("F:\\FFF\\respiratory-sound-database\\Respiratory_Sound_Database\\Respiratory_Sound_Database/patient_diagnosis.csv", header=None)
labels = np.array([p_diag[p_diag[0] == x][1].values[0] for x in p_id_in_file])

# Remove specific labels
features = np.delete(features, np.where((labels == 'Asthma') | (labels == 'LRTI'))[0], axis=0)
labels = np.delete(labels, np.where((labels == 'Asthma') | (labels == 'LRTI'))[0], axis=0)

# Preprocess labels
le = LabelEncoder()
i_labels = le.fit_transform(labels)
oh_labels = to_categorical(i_labels)
features = features.reshape(*features.shape, 1)

# Train-test split
x_train, x_test, y_train, y_test = train_test_split(features, oh_labels, stratify=oh_labels, test_size=0.2, random_state=42)

# =============================================================================
# DEFINE CNN MODEL
# =============================================================================
cnn_model = Sequential([
    Input(shape=(max_pad_len, 1)),
    Conv1D(16, 2, activation='relu'),
    MaxPooling1D(2),
    Dropout(0.2),
    Conv1D(32, 2, activation='relu'),
    MaxPooling1D(2),
    Dropout(0.2),
    Conv1D(64, 2, activation='relu'),
    MaxPooling1D(2),
    Dropout(0.2),
    Conv1D(128, 2, activation='relu'),
    MaxPooling1D(2),
    Dropout(0.2),
    GlobalAveragePooling1D()
])

cnn_model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
cnn_model.summary()

# =============================================================================
# FEATURE EXTRACTION
# =============================================================================
x_train_features = cnn_model.predict(x_train)
x_test_features = cnn_model.predict(x_test)
x_train_flat = x_train_features.reshape(x_train_features.shape[0], -1)
x_test_flat = x_test_features.reshape(x_test_features.shape[0], -1)

# =============================================================================
# t-SNE/UMAP VISUALIZATION (Fixed Subset Logic)
# =============================================================================
all_features = np.vstack([x_train_flat, x_test_flat])
all_labels = np.concatenate([np.argmax(y_train, axis=1), np.argmax(y_test, axis=1)])

# Fix: Dynamic subset size
total_samples = all_features.shape[0]
subset_size = min(1000, total_samples)  # Automatically adjust subset size
print(f"Using {subset_size} samples for visualization")

np.random.seed(42)
subset_indices = np.random.choice(total_samples, subset_size, replace=False)
subset_features = all_features[subset_indices]
subset_labels = all_labels[subset_indices]

# t-SNE
tsne = TSNE(n_components=2, perplexity=30, random_state=42)
tsne_results = tsne.fit_transform(subset_features)

# UMAP
umap_reducer = umap.UMAP(n_neighbors=15, min_dist=0.1, random_state=42)
umap_results = umap_reducer.fit_transform(subset_features)

# Plot
plt.figure(figsize=(18, 6))
plt.subplot(1, 2, 1)
sns.scatterplot(x=tsne_results[:,0], y=tsne_results[:,1], hue=le.inverse_transform(subset_labels), palette='tab10')
plt.title('t-SNE Visualization')

plt.subplot(1, 2, 2)
sns.scatterplot(x=umap_results[:,0], y=umap_results[:,1], hue=le.inverse_transform(subset_labels), palette='tab10')
plt.title('UMAP Visualization')
plt.show()

# =============================================================================
# XGBOOST MODELING
# =============================================================================
y_train_int = np.argmax(y_train, axis=1)
y_test_int = np.argmax(y_test, axis=1)

class_weights = {
    0: 10,  # Bronchiectasis
    1: 10,  # Bronchiolitis
    2: 1,   # COPD
    3: 5,   # Healthy
    4: 5,   # Pneumonia
    5: 3    # URTI
}

xgb_model = xgb.XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    gamma=0.1,
    scale_pos_weight=5,
    objective='multi:softprob',
    num_class=len(le.classes_),
    eval_metric='mlogloss',
    early_stopping_rounds=10
)

xgb_model.fit(
    x_train_flat, y_train_int,
    eval_set=[(x_test_flat, y_test_int)],
    verbose=True
)

# =============================================================================
# EVALUATION
# =============================================================================
y_pred = xgb_model.predict(x_test_flat)
print(f"Test Accuracy: {accuracy_score(y_test_int, y_pred):.4f}")
print("Classification Report:")
print(classification_report(y_test_int, y_pred, target_names=le.classes_, zero_division=1))
print("Confusion Matrix:")
print(confusion_matrix(y_test_int, y_pred))
```
